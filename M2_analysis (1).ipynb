```python
import pandas as pd

# Load dataset
df = pd.read_csv("online_retail_II.csv", encoding="ISO-8859-1")
df["Invoice"] = df["Invoice"].astype(str)

# Basic cleaning
df = df.dropna(subset=["Invoice", "Description", "Country"])
df = df[~df["Invoice"].str.startswith("C")]   # remove cancellations
df = df[df["Quantity"] > 0]                   # valid quantity
df = df[df["Price"] > 0]                      # valid price

# Reduce size (preprocessing decision)
df = df[df["Country"] == "United Kingdom"]    # biggest portion, reduces RAM a lot

TOP_N_ITEMS = 200
top_items = df["Description"].value_counts().head(TOP_N_ITEMS).index
df = df[df["Description"].isin(top_items)]

MAX_INVOICES = 8000
keep_invoices = df["Invoice"].drop_duplicates().head(MAX_INVOICES)
df = df[df["Invoice"].isin(keep_invoices)]

print("Filtered rows:", df.shape[0])
print("Invoices:", df["Invoice"].nunique())
print("Unique products:", df["Description"].nunique())
df.head()
```

```python
basket = (
    df.groupby(["Invoice", "Description"])["Quantity"]
      .sum()
      .unstack(fill_value=0)
)

basket = basket.gt(0)

print("Basket shape:", basket.shape)
basket.head()
```

```python
!pip -q install mlxtend
from mlxtend.frequent_patterns import apriori, association_rules

frequent_items = apriori(basket, min_support=0.02, use_colnames=True)
rules = association_rules(frequent_items, metric="lift", min_threshold=1.0)
rules = rules.sort_values(["lift", "confidence"], ascending=False)

print("Frequent itemsets:", len(frequent_items))
print("Rules:", len(rules))
rules.head(10)
```

```python
rules.head(20).to_csv("top_rules.csv", index=False)
print("Saved top_rules.csv")
```
